```python
# 1. Import các thư viện cần thiết
import numpy as np
import pandas as pd
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
from gensim.utils import simple_preprocess
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

```python
# ────────────────────────────────────────────────
# 2. Định nghĩa class Doc2VecVectorizer (để dùng trong Pipeline)
# ────────────────────────────────────────────────
class Doc2VecVectorizer(BaseEstimator, TransformerMixin):
    """
    Transformer để chuyển text → Doc2Vec vector
    Dùng infer_vector cho cả train và test
    """
    def __init__(self, model):
        self.model = model

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        return np.array([
            self.model.infer_vector(
                simple_preprocess(text.lower())   # ← XÓA max_vocab_size ở đây
            )
            for text in X
        ])
```

```python
# ────────────────────────────────────────────────
# 3. Chuẩn bị dữ liệu TaggedDocument để TRAIN Doc2Vec
# ────────────────────────────────────────────────
# Giả sử X_train là list[str] hoặc pd.Series
documents = [
    TaggedDocument(
        words=simple_preprocess(text.lower()),   # thay bằng word_tokenize nếu dùng underthesea
        tags=[i]
    )
    for i, text in enumerate(X_train)
]
```

```python
# ────────────────────────────────────────────────
# 4. Huấn luyện Doc2Vec model
# ────────────────────────────────────────────────
print("Đang huấn luyện Doc2Vec...")

d2v_model = Doc2Vec(
    documents=documents,
    vector_size=200,          # 128–300 thường tốt, 200 là lựa chọn cân bằng
    window=5,                 # ngữ cảnh gần
    min_count=2,              # bỏ từ quá hiếm
    workers=4,                # số luồng (tùy CPU)
    dm=1,                     # 1 = PV-DM (thường tốt hơn cho phân loại), 0 = PV-DBOW
    epochs=30,                # 20–60 tùy lượng dữ liệu (ít dữ liệu thì tăng epochs)
    negative=5,               # negative sampling
    hs=0                      # 0 = negative sampling, 1 = hierarchical softmax
)

print("Doc2Vec đã huấn luyện xong. Số từ trong vocab:", len(d2v_model.wv))
```

```python
# ────────────────────────────────────────────────
# 5. Tạo Pipeline hoàn chỉnh
# ────────────────────────────────────────────────
from sklearn.neighbors import KNeighborsClassifier
pipeline = Pipeline([
    ('doc2vec',  Doc2VecVectorizer(d2v_model)),
    ('scaler',   StandardScaler()),               # rất quan trọng với SVM
    ('clf',      KNeighborsClassifier(
        n_neighbors=15,             # thường thử 5–30, 10–20 hay gặp nhất
        weights='distance',         # 'uniform' hoặc 'distance' (thường distance tốt hơn)
        metric='cosine',            # 'euclidean', 'manhattan', 'cosine' → cosine thường tốt cho embedding
        algorithm='auto',           # hoặc 'ball_tree' nếu dữ liệu lớn
        n_jobs=-1,
        p=2                         # chỉ dùng khi metric=euclidean
    ))
])
```

```python
# ────────────────────────────────────────────────
# 6. Huấn luyện toàn bộ pipeline
# ────────────────────────────────────────────────
print("Đang fit pipeline...")
pipeline.fit(X_train, y_train)
print("Pipeline đã fit xong!")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np
import torch
import textwrap   # để in đẹp hơn nếu cần

y_pred = pipeline.predict(X_test)
# ----------------------------------------------------------------------
# Đảm bảo y_test và y_pred ở dạng numpy array (phổ biến nhất)
# ----------------------------------------------------------------------
if isinstance(y_test, torch.Tensor):
    y_test_np    = y_test.cpu().numpy()
else:
    y_test_np    = np.asarray(y_test)

if isinstance(y_pred, torch.Tensor):
    y_pred_np    = y_pred.cpu().numpy()
else:
    y_pred_np    = np.asarray(y_pred)

# Nếu là multi-class và y_pred là probability → lấy argmax
if len(y_pred_np.shape) > 1 and y_pred_np.shape[1] > 1:
    y_pred_np = np.argmax(y_pred_np, axis=1)

# ----------------------------------------------------------------------
# Tính các metric cơ bản
# ----------------------------------------------------------------------
accuracy = accuracy_score(y_test_np, y_pred_np)

# In header
print("\n" + "="*80)
print("              MODEL EVALUATION ON TEST SET")
print("="*80)

# Accuracy chi tiết
print(f"\nOverall Accuracy:")
print(f"  • {accuracy:.4f}  ({accuracy*100:.2f}%)")
print("-"*60)

# Classification Report (chi tiết, 4 chữ số thập phân)
print("\nClassification Report:")
print(classification_report(
    y_test_np,
    y_pred_np,
    digits=4,
    zero_division=0,
    target_names=[str(c) for c in pipeline.classes_]   # nếu có class names đẹp
    # target_names = ['Negative', 'Positive']  # → bạn có thể hard-code tên lớp nếu muốn
))

# ----------------------------------------------------------------------
# Confusion Matrix - cả dạng số lượng và dạng chuẩn hóa (%)
# ----------------------------------------------------------------------
cm = confusion_matrix(y_test_np, y_pred_np)

print("\nConfusion Matrix (Raw counts):")
print(cm)

# Chuẩn hóa theo hàng (precision-oriented: tỷ lệ dự đoán đúng trong mỗi lớp thật)
cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
cm_normalized = np.nan_to_num(cm_normalized)  # tránh NaN nếu có lớp không có mẫu

# ----------------------------------------------------------------------
# Một số metric tổng hợp bổ sung (macro / weighted)
# ----------------------------------------------------------------------
print("\nSummary Metrics:")
print(f"  Macro    Avg Precision : {precision_score(y_test_np, y_pred_np, average='macro', zero_division=0):.4f}")
print(f"  Macro    Avg Recall    : {recall_score(y_test_np, y_pred_np, average='macro', zero_division=0):.4f}")
print(f"  Macro    Avg F1-score  : {f1_score(y_test_np, y_pred_np, average='macro', zero_division=0):.4f}")
print(f"  Weighted Avg F1-score : {f1_score(y_test_np, y_pred_np, average='weighted', zero_division=0):.4f}")

# Kết thúc đẹp
# ----------------------------------------------------------------------
torch.cuda.empty_cache() if torch.cuda.is_available() else None

print("\n" + "="*80)
print(f"HOÀN TẤT ĐÁNH GIÁ!  •  Accuracy = {accuracy*100:.2f}%")
print("="*80 + "\n")
```
